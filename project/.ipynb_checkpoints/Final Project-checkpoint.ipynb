```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
# Rotation-curve toy models

def v_disk(r, v0=180.0, Rd=5.0):
    """
    Toy exponential disk rotation curve (km/s).
    - Rises in the inner galaxy, then declines outside the disk.
    """
    r = np.asarray(r, dtype=float)
    return v0 * (r / Rd) * np.exp(-0.5 * r / Rd)

def v_halo_cored_iso(r, v_inf=150.0, r_c=10.0):
    """
    Cored isothermal-like halo rotation curve (km/s).
    - Roughly linear at small r, approaches flat v_inf at large r.
    """
    r = np.asarray(r, dtype=float)
    return v_inf * r / np.sqrt(r**2 + r_c**2)

def v_total(r, disk_params, halo_params):
    """
    Total circular speed via quadrature sum:
        v_tot^2 = v_disk^2 + v_halo^2
    """
    v_d = v_disk(r, *disk_params)
    v_h = v_halo_cored_iso(r, *halo_params)
    return np.sqrt(v_d**2 + v_h**2)
# Cell 1: Define the rotation-curve models.
# In this cell I import the libraries and define simple functions for a galaxy’s rotation speed from (1) a luminous exponential disk,
# (2) a cored dark-matter halo, and (3) the combined total rotation curve using a quadrature sum.
```

```python
# Synthetic data (mock observed rotation curve)

r_grid = np.linspace(0.1, 60.0, 400)  # kpc, smooth curve radii

disk_true = (180.0, 5.0)   # (v0, Rd)
halo_true = (150.0, 10.0)  # (v_inf, r_c)

v_disk_grid  = v_disk(r_grid, *disk_true)
v_halo_grid  = v_halo_cored_iso(r_grid, *halo_true)
v_tot_grid   = np.sqrt(v_disk_grid**2 + v_halo_grid**2)

# "Observed" sample points + noise
rng = np.random.default_rng(42)
r_data  = np.linspace(3.0, 55.0, 18)
sigma_v = 5.0 * np.ones_like(r_data) # constant error bars (km/s)
v_true  = v_total(r_data, disk_true, halo_true)
v_obs   = v_true + rng.normal(0.0, sigma_v)

# Cell 2: Generate synthetic (mock) rotation-curve observations.
# Here I choose example parameters for the disk and halo, compute smooth model curves on a radius grid, 
# then sample the curve at a few radii and add Gaussian noise + error bars to mimic real observational uncertainty.
```

```python
plt.figure(figsize=(8, 5))

plt.plot(r_grid, v_disk_grid, "--", label="Disk-only (luminous)")
plt.plot(r_grid, v_tot_grid,  "-",  label="Disk + Halo")
plt.plot(r_grid, v_halo_grid, ":",  label="Halo contribution")

plt.errorbar(r_data, v_obs, yerr=sigma_v, fmt="o", ms=4, capsize=3,
             label="Mock observed data")

plt.xlabel("Radius r (kpc)")
plt.ylabel("Rotation speed v (km/s)")
plt.title("Synthetic Rotation Curve: Disk-only vs Disk+Halo")
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# Cell 3: Plot disk-only vs disk+halo and compare to the mock data.
# This plot visually demonstrates the key idea: a disk-only model peaks and declines, 
# while adding a halo keeps the rotation curve flatter at large radii.
# I also overlay the mock observations with error bars to show what the “data” supports.
```

```python
def chi2(y_model, y_data, sigma):
    """Chi-square = sum((data-model)^2 / sigma^2)."""
    return np.sum((y_data - y_model)**2 / sigma**2)

# Model predictions evaluated at the data radii
v_model_disk  = v_disk(r_data, *disk_true)
v_model_total = v_total(r_data, disk_true, halo_true)

chi2_disk  = chi2(v_model_disk,  v_obs, sigma_v)
chi2_total = chi2(v_model_total, v_obs, sigma_v)

print(f"chi^2 (disk only)  = {chi2_disk:.2f}")
print(f"chi^2 (disk+halo)  = {chi2_total:.2f}")

# Cell 4: Quantify the model fit using chi-square.
# In this cell I compute the chi-square statistic for disk-only and disk+halo models against the same mock data. 
```

```python
# SPARC Burkert fit table: picked galaxy (CamB)

cols = [
    "Name","Ydisk","e_Ydisk","Ybul","e_Ybul",
    "D","e_D","inc","e_inc",
    "V200","e_V200","C200","e_C200",
    "rs","e_rs","log_rhos","e_log_rhos",
    "log_M200","e_log_M200","Chi"
]

fname = "parameter_Burkert.mrt"
try:
    burkert = pd.read_fwf(fname, comment="#", names=cols)
except Exception:
    burkert = pd.read_table(fname, delim_whitespace=True, comment="#", header=None, names=cols)

burkert["Name"] = burkert["Name"].astype(str).str.strip()
camB = burkert.loc[burkert["Name"] == "CamB"].iloc[0]

print(camB[["Name","D","inc","V200","rs","log_rhos","log_M200","Chi"]])

# Cell 5: Load real parameter-fit results (SPARC Burkert table) and select one galaxy.
# This cell reads the Burkert halo fit table (from SPARC) and extracts the row for the galaxy CamB. 
# These parameters come from fitting real galaxies and help anchor my toy halo parameters to something realistic.
```

```python
# mapping to my toy halo parameters
V200_camb = float(camB["V200"]) # km/s
rs_camb   = float(camB["rs"]) # kpc

v_inf_camb  = V200_camb / 4.0
r_core_camb = rs_camb / 3.0

print(f"CamB-anchored toy halo: v_inf={v_inf_camb:.2f} km/s, r_c={r_core_camb:.2f} kpc")

# Cell 6: Convert CamB fit parameters into toy halo settings.
# Here I map the CamB fit parameters into the simplified halo model I’m using. 
# This conversion is not physically exact, but it gives reasonable halo values to demonstrate the effect of a halo on rotation curves.
```

```python
r_grid = np.linspace(0.1, 60.0, 400)

disk_params = (180.0, 5.0)
halo_params = (v_inf_camb, r_core_camb)

v_disk_grid = v_disk(r_grid, *disk_params)
v_halo_grid = v_halo_cored_iso(r_grid, *halo_params)
v_tot_grid  = np.sqrt(v_disk_grid**2 + v_halo_grid**2)

plt.figure(figsize=(8, 5))
plt.plot(r_grid, v_disk_grid, "--", label="Disk-only (luminous)")
plt.plot(r_grid, v_tot_grid,  "-",  label="Disk + Halo (CamB-anchored)")
plt.plot(r_grid, v_halo_grid, ":",  label="Halo contribution")

plt.xlabel("Radius r (kpc)")
plt.ylabel("Rotation speed v (km/s)")
plt.title("Rotation Curves (CamB-anchored toy parameters)")
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()


# Cell 7: Plot a CamB-anchored disk+halo rotation curve.
# Using the CamB-based halo scale, I re-plot the disk-only, halo-only, and disk+halo curves. 
# This shows how a halo component changes the outer rotation behavior when using a real-galaxy-inspired parameter scale.
```

```python
from matplotlib.animation import FuncAnimation, PillowWriter

# Animation: compare disk-only vs disk+halo

disk_params = (180.0, 5.0)
halo_params = (v_inf_camb, r_core_camb)

rng = np.random.default_rng(0)
N = 1200
Rmax = 30.0 # displayed radius range (kpc)
Rd_vis = 6.0

# Star "particles" initial conditions
r = rng.exponential(scale=Rd_vis, size=N)
r = np.clip(r, 0.3, Rmax)
theta0 = rng.uniform(0, 2*np.pi, size=N)

# Rotation speeds at each star radius
v_disk_r = v_disk(r, *disk_params)
v_halo_r = v_halo_cored_iso(r, *halo_params)
v_tot_r  = np.sqrt(v_disk_r**2 + v_halo_r**2)

# Convert to angular speed omega = v/r (in 1/s) to spin the points
KPC_IN_KM = 3.085677581e16
eps = 1e-6
omega_disk = v_disk_r / ((r + eps) * KPC_IN_KM)
omega_tot  = v_tot_r  / ((r + eps) * KPC_IN_KM)

# Time step per frame (purely for visualization)
dt = 8e13 # seconds per frame (~2.5 million years)
frames = 180

# Initial positions
x0 = r * np.cos(theta0)
y0 = r * np.sin(theta0)

fig, axes = plt.subplots(1, 2, figsize=(10, 5), constrained_layout=True)
ax1, ax2 = axes

for ax, title in zip(axes, ["Disk-only (luminous)", "Disk + halo (dark matter)"]):
    ax.set_title(title)
    ax.set_aspect("equal")
    ax.set_xlim(-Rmax, Rmax)
    ax.set_ylim(-Rmax, Rmax)
    ax.set_xlabel("x (kpc)")
    ax.set_ylabel("y (kpc)")
    for rr in [5, 10, 20, 30]:
        ax.add_patch(plt.Circle((0, 0), rr, fill=False, alpha=0.08))

# Color by radius so you can see inner vs outer
sc1 = ax1.scatter(x0, y0, s=6, c=r, cmap="viridis", alpha=0.85)
sc2 = ax2.scatter(x0, y0, s=6, c=r, cmap="viridis", alpha=0.85)

def update(i):
    t = i * dt
    th1 = theta0 + omega_disk * t
    th2 = theta0 + omega_tot  * t

    x1 = r * np.cos(th1); y1 = r * np.sin(th1)
    x2 = r * np.cos(th2); y2 = r * np.sin(th2)

    sc1.set_offsets(np.column_stack([x1, y1]))
    sc2.set_offsets(np.column_stack([x2, y2]))
    return sc1, sc2

anim = FuncAnimation(fig, update, frames=frames, interval=40, blit=True)

out = "rotation_compare.gif"
anim.save(out, writer=PillowWriter(fps=25))
plt.close(fig)

print("Saved:", out)


# Cell 8: Create an animation showing differential rotation with/without a halo.
# In this cell I simulate star “particles” at different radii and update their angles over time using angular speed, 
# The left panel uses disk-only speeds, causing strong differential rotation (shear), while the right panel includes a halo, 
# producing a flatter rotation curve and more coherent outer rotation. The output is saved as a gif.
```
